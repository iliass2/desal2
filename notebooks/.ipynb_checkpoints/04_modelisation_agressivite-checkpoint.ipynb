# 04 — Modélisation de l’agressivité / corrosivité

## Objectif

Prédire la cible `Larson_corrosive` en distinguant deux questions scientifiques :

- **Expérience A — reproduction de la règle** : le modèle reçoit `Cl`, `SO4` et `HCO3`, qui définissent directement Larson.
- **Expérience B — prédiction indirecte** : ces trois variables sont retirées afin de tester si les autres paramètres contiennent suffisamment d’information pour anticiper le risque corrosif.

Cette séparation évite de présenter un score très élevé de l’Expérience A comme une découverte indépendante du machine learning.

## 1. Configuration

```python
from pathlib import Path
import sys
import warnings

warnings.filterwarnings("ignore")

PROJECT_ROOT = (
    Path.cwd().resolve().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd().resolve()
)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
FIGURES = PROJECT_ROOT / "reports" / "figures"
RESULTS = PROJECT_ROOT / "reports" / "results"
MODELS = PROJECT_ROOT / "models"

for directory in [DATA_PROCESSED, FIGURES, RESULTS, MODELS]:
    directory.mkdir(parents=True, exist_ok=True)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RANDOM_STATE = 42
pd.set_option("display.max_columns", 100)
```

```python
from sklearn.model_selection import train_test_split
import joblib

from src.data_cleaning import basic_cleaning, load_water_quality
from src.evaluation import (
    classification_metrics,
    cross_validation_table,
    save_confusion_matrix,
    save_precision_recall_curve,
    save_roc_curve,
)
from src.indices_chimiques import add_larson_features
from src.interpretability import permutation_importance_table
from src.modeling import classification_models, needs_scaling
from src.preprocessing import make_model_pipeline
```

## 2. Construction du jeu de données

```python
path = DATA_RAW / "water_quality.xlsx"

mhl = add_larson_features(
    basic_cleaning(load_water_quality(path, "MHLATHUZE"))
)
luv = add_larson_features(
    basic_cleaning(load_water_quality(path, "LUVUVU"))
)

common = set(mhl.columns) & set(luv.columns)
all_features = [
    column
    for column in ["pH", "EC", "TDS", "Na", "K", "Ca", "Mg", "Cl", "HCO3", "SO4", "NO3"]
    if column in common
]

data = pd.concat(
    [
        mhl[all_features + ["Larson_corrosive"]],
        luv[all_features + ["Larson_corrosive"]],
    ],
    ignore_index=True,
)

data = (
    data.replace([np.inf, -np.inf], np.nan)
    .dropna(subset=["Larson_corrosive"])
    .reset_index(drop=True)
)
data["Larson_corrosive"] = data["Larson_corrosive"].astype(int)

print("Dimensions :", data.shape)
display(data["Larson_corrosive"].value_counts().to_frame("effectif"))
display((100 * data["Larson_corrosive"].value_counts(normalize=True)).round(2).to_frame("pourcentage"))
```

### Interprétation

La cible n’est pas une observation indépendante : elle est dérivée de Larson. Cette caractéristique détermine la manière dont les expériences doivent être interprétées.

## 3. Définition des deux expériences

```python
direct_larson_features = ["Cl", "SO4", "HCO3"]

features_a = all_features.copy()
features_b = [column for column in all_features if column not in direct_larson_features]

print("Expérience A — toutes les variables :", features_a)
print("Expérience B — sans variables directes de Larson :", features_b)
```

### Interprétation

**Expérience A** mesure surtout la capacité du modèle à approximer une règle chimique connue. **Expérience B** est plus ambitieuse : elle cherche un signal indirect dans les autres paramètres physico-chimiques.

## 4. Split train/test commun aux deux expériences

```python
indices = np.arange(len(data))
train_idx, test_idx = train_test_split(
    indices,
    test_size=0.20,
    stratify=data["Larson_corrosive"],
    random_state=RANDOM_STATE,
)

y_train = data.loc[train_idx, "Larson_corrosive"]
y_test = data.loc[test_idx, "Larson_corrosive"]

print("Train :", len(train_idx), "Test :", len(test_idx))
display(y_train.value_counts(normalize=True).round(4).to_frame("proportion_train"))
display(y_test.value_counts(normalize=True).round(4).to_frame("proportion_test"))
```

### Interprétation

Les deux expériences utilisent exactement les mêmes observations de train et de test. Ainsi, la différence de performance peut être attribuée au retrait de `Cl`, `SO4` et `HCO3`, et non à un changement de découpage.

## 5. Fonction d’évaluation commune

```python
def run_experiment(label, feature_list):
    X_train = data.loc[train_idx, feature_list]
    X_test = data.loc[test_idx, feature_list]

    models = classification_models(RANDOM_STATE)
    pipelines = {}
    cv_rows = []

    for name, model in models.items():
        pipeline = make_model_pipeline(
            model=model,
            numeric_features=feature_list,
            scale=needs_scaling(name),
        )
        pipelines[name] = pipeline
        cv_table = cross_validation_table(
            pipeline,
            X_train,
            y_train,
            n_splits=5,
            random_state=RANDOM_STATE,
        )
        for _, row in cv_table.iterrows():
            cv_rows.append({"experiment": label, "model": name, **row.to_dict()})

    cv_results = pd.DataFrame(cv_rows)
    cv_pivot = cv_results.pivot(index="model", columns="metric", values="mean")

    # Sélection uniquement à partir de la validation croisée du train.
    non_dummy_cv = cv_pivot.drop(index="Dummy Classifier", errors="ignore").copy()
    ranking_metrics = ["f1", "roc_auc", "pr_auc", "mcc"]
    cv_ranking = non_dummy_cv.sort_values(ranking_metrics, ascending=False)
    best_name = cv_ranking.index[0]
    best_model = pipelines[best_name]

    # Le test n'est consulté qu'après la sélection du modèle.
    best_model.fit(X_train, y_train)
    final_metrics = {
        "experiment": label,
        "model": best_name,
        **classification_metrics(best_model, X_test, y_test),
    }
    test_results = pd.DataFrame([final_metrics])

    importance = permutation_importance_table(
        best_model,
        X_test,
        y_test,
        scoring="f1",
    )

    return {
        "label": label,
        "features": feature_list,
        "X_test": X_test,
        "cv": cv_results,
        "cv_ranking": cv_ranking,
        "test": test_results,
        "best_name": best_name,
        "best_model": best_model,
        "importance": importance,
    }
```

## 6. Expérience A — avec `Cl`, `SO4`, `HCO3`

Dans chaque expérience, les algorithmes sont comparés par validation croisée sur le train. Le meilleur modèle est sélectionné selon F1, puis ROC-AUC, PR-AUC et MCC. **Le jeu de test n’intervient pas dans la sélection** et sert uniquement à l’évaluation finale du modèle retenu.

```python
exp_a = run_experiment("A_with_larson_inputs", features_a)

print("Classement CV — Expérience A :")
display(exp_a["cv_ranking"].round(4))
print("Meilleur modèle A sélectionné par CV :", exp_a["best_name"])
print("Évaluation finale test — Expérience A :")
display(exp_a["test"].round(4))
```

### Interprétation de l’Expérience A

Une performance élevée est attendue parce que le modèle observe directement les variables qui déterminent la cible. Cette expérience est utile comme **contrôle de cohérence** et comme démonstration de reproduction prédictive de l’indice, mais elle ne doit pas être surinterprétée.

## 7. Expérience B — sans `Cl`, `SO4`, `HCO3`

```python
exp_b = run_experiment("B_without_larson_inputs", features_b)

print("Classement CV — Expérience B :")
display(exp_b["cv_ranking"].round(4))
print("Meilleur modèle B sélectionné par CV :", exp_b["best_name"])
print("Évaluation finale test — Expérience B :")
display(exp_b["test"].round(4))
```

### Interprétation de l’Expérience B

Cette expérience est la plus intéressante scientifiquement. Une baisse de performance par rapport à A est normale et informative : elle quantifie la dépendance de la cible envers les trois ions retirés. Une performance encore supérieure à la baseline indiquerait qu’un signal indirect existe dans les autres paramètres.

## 8. Comparaison directe A vs B

```python
comparison = pd.concat([exp_a["test"], exp_b["test"]], ignore_index=True)
display(comparison.round(4))

# Une seule ligne finale par expérience : le modèle a déjà été choisi par CV.
best_comparison = comparison.copy()
display(best_comparison.round(4))
```

### Interprétation

La comparaison A/B est une **analyse d’ablation** entre deux expériences définies à l’avance. Chaque modèle a été sélectionné sur le train uniquement ; les performances test servent donc à quantifier l’effet du retrait de `Cl`, `SO4` et `HCO3`, sans utiliser le test pour choisir l’algorithme.

## 9. Figures des meilleurs modèles

```python
from IPython.display import Image, display

for prefix, exp in [("aggressiveness_A", exp_a), ("aggressiveness_B", exp_b)]:
    model = exp["best_model"]
    name = exp["best_name"]
    X_test = exp["X_test"]

    paths = {
        "cm": FIGURES / f"{prefix}_confusion_matrix.png",
        "cm_norm": FIGURES / f"{prefix}_confusion_matrix_normalized.png",
        "roc": FIGURES / f"{prefix}_roc_curve.png",
        "pr": FIGURES / f"{prefix}_precision_recall_curve.png",
    }

    save_confusion_matrix(model, X_test, y_test, f"{prefix} — {name}", paths["cm"])
    save_confusion_matrix(
        model,
        X_test,
        y_test,
        f"{prefix} normalisée — {name}",
        paths["cm_norm"],
        normalize="true",
    )
    save_roc_curve(model, X_test, y_test, f"{prefix} — {name}", paths["roc"])
    save_precision_recall_curve(model, X_test, y_test, f"{prefix} — {name}", paths["pr"])

    print()
    print(f"{prefix} — {name}")
    for path in paths.values():
        display(Image(filename=str(path)))
```

## 10. Importance des variables

```python
for label, exp in [("Expérience A", exp_a), ("Expérience B", exp_b)]:
    print(label)
    display(exp["importance"])

    ax = exp["importance"].sort_values("importance_mean").plot.barh(
        x="feature",
        y="importance_mean",
        xerr="importance_std",
        figsize=(8, 5),
        legend=False,
        title=f"Permutation importance — {label}",
    )
    ax.set_xlabel("Baisse moyenne du F1 après permutation")
    plt.tight_layout()
    plt.show()
```

### Interprétation

Dans l’Expérience A, une forte importance de `Cl`, `SO4` ou `HCO3` est mécaniquement cohérente avec la définition de Larson. Dans B, les variables importantes représentent uniquement des associations indirectes ; elles ne doivent pas être interprétées comme des causes chimiques sans validation supplémentaire.

## 11. Sauvegarde

```python
for suffix, exp in [("A", exp_a), ("B", exp_b)]:
    exp["cv"].to_csv(RESULTS / f"aggressiveness_{suffix}_cv_metrics.csv", index=False)
    exp["test"].to_csv(RESULTS / f"aggressiveness_{suffix}_test_metrics.csv", index=False)
    exp["importance"].to_csv(RESULTS / f"aggressiveness_{suffix}_feature_importance.csv", index=False)
    joblib.dump(exp["best_model"], MODELS / f"aggressiveness_{suffix}_best_model.pkl")

comparison.to_csv(RESULTS / "aggressiveness_A_vs_B_test_metrics.csv", index=False)
best_comparison.to_csv(RESULTS / "aggressiveness_A_vs_B_best_models.csv", index=False)
print("Résultats des deux expériences sauvegardés.")
```

## 12. Conclusion du notebook

- L’Expérience A évalue la capacité des modèles à reproduire une règle construite à partir de variables disponibles.
- L’Expérience B retire volontairement les trois variables constitutives de Larson et constitue l’analyse Data Science la plus informative.
- Pour A comme pour B, le meilleur algorithme est sélectionné exclusivement par validation croisée sur le train.
- Le test est ensuite utilisé une seule fois par expérience pour l’évaluation finale du modèle retenu.
- L’écart entre A et B est interprété comme une analyse d’ablation et non comme une compétition de modèles sur le test.
