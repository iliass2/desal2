# 05 — Synthèse et discussion

## Objectif

Rassembler les résultats produits par les notebooks 03 et 04, comparer la potabilité aux deux expériences d’agressivité et formuler une lecture scientifique prudente adaptée au mémoire de PFE.

## 1. Configuration

```python
from pathlib import Path
import sys
import warnings

warnings.filterwarnings("ignore")

PROJECT_ROOT = (
    Path.cwd().resolve().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd().resolve()
)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
FIGURES = PROJECT_ROOT / "reports" / "figures"
RESULTS = PROJECT_ROOT / "reports" / "results"
MODELS = PROJECT_ROOT / "models"

for directory in [DATA_PROCESSED, FIGURES, RESULTS, MODELS]:
    directory.mkdir(parents=True, exist_ok=True)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RANDOM_STATE = 42
pd.set_option("display.max_columns", 100)
```

## 2. Chargement des résultats sauvegardés

```python
result_files = {
    "Potabilité — CV": RESULTS / "potability_cv_metrics.csv",
    "Potabilité — Test": RESULTS / "potability_test_metrics.csv",
    "Agressivité A — CV": RESULTS / "aggressiveness_A_cv_metrics.csv",
    "Agressivité A — Test": RESULTS / "aggressiveness_A_test_metrics.csv",
    "Agressivité B — CV": RESULTS / "aggressiveness_B_cv_metrics.csv",
    "Agressivité B — Test": RESULTS / "aggressiveness_B_test_metrics.csv",
}

loaded = {}
for label, path in result_files.items():
    if path.exists():
        loaded[label] = pd.read_csv(path)
        print()
        print(label)
        display(loaded[label].round(4))
    else:
        print(f"{label} : absent — exécuter les notebooks 03 et 04.")
```

## 3. Meilleurs modèles de chaque axe

```python
summary_rows = []
for axis, key in [
    ("Potabilité", "Potabilité — Test"),
    ("Agressivité A — avec variables Larson", "Agressivité A — Test"),
    ("Agressivité B — sans variables Larson", "Agressivité B — Test"),
]:
    if key in loaded and not loaded[key].empty:
        # Les fichiers test contiennent déjà uniquement le modèle sélectionné par CV.
        row = loaded[key].iloc[0].to_dict()
        row["axe"] = axis
        summary_rows.append(row)

summary = pd.DataFrame(summary_rows)
if not summary.empty:
    columns = [
        column
        for column in [
            "axe", "model", "accuracy", "precision", "recall", "f1",
            "roc_auc", "pr_auc", "mcc"
        ]
        if column in summary.columns
    ]
    display(summary[columns].round(4))
    summary[columns].to_csv(RESULTS / "synthese_modeles.csv", index=False)
```

### Interprétation

Ce tableau ne sélectionne plus de modèle à partir des performances test. Il rassemble uniquement les évaluations finales des modèles **déjà sélectionnés par validation croisée sur les jeux d’entraînement** dans les notebooks 03 et 04.

## 5. Lecture scientifique de l’agressivité

### Expérience A

Le modèle reçoit `Cl`, `SO4` et `HCO3`, alors que la cible `Larson_corrosive` est calculée à partir de ces mêmes ions. Une performance élevée signifie principalement que le modèle apprend à reproduire la règle de Larson.

### Expérience B

Les trois variables directes sont retirées. Cette expérience cherche à savoir si les autres paramètres physico-chimiques contiennent un signal indirect permettant d’anticiper la corrosivité. Une baisse des métriques par rapport à A est attendue et constitue une information scientifique importante.

### Analyse d’ablation

Le contraste entre A et B quantifie le coût du retrait de l’information qui définit directement la cible. C’est cette comparaison qui rend la partie « agressivité » plus défendable devant un jury de Data Science.

## 6. Comparaison A vs B

```python
if {"Agressivité A — Test", "Agressivité B — Test"}.issubset(loaded):
    a_best = loaded["Agressivité A — Test"].iloc[0]
    b_best = loaded["Agressivité B — Test"].iloc[0]

    metrics = ["accuracy", "precision", "recall", "f1", "roc_auc", "pr_auc", "mcc"]
    delta = pd.DataFrame({
        "metric": metrics,
        "A": [a_best.get(metric) for metric in metrics],
        "B": [b_best.get(metric) for metric in metrics],
    })
    delta["B_minus_A"] = delta["B"] - delta["A"]
    display(delta.round(4))
    delta.to_csv(RESULTS / "aggressiveness_ablation_delta.csv", index=False)
```

### Interprétation

Une différence négative `B_minus_A` indique une perte de performance après retrait des variables constitutives de Larson. Cette perte ne constitue pas un échec du projet : elle démontre au contraire quantitativement à quel point la cible dépend de l’information chimique directe.

## 7. Mise en perspective avec le dessalement

Dans une station de dessalement, un système de Data Science réellement opérationnel devrait idéalement recevoir des mesures locales horodatées, connaître l’étape du procédé à laquelle chaque échantillon est prélevé et être validé sur des périodes distinctes.

Dans ce PFE, la station d’Al Hoceima sert de **contexte d’application et de discussion**. Les datasets publics utilisés pour entraîner les modèles ne doivent pas être présentés comme des mesures locales de la station.

## 8. Limites du projet

1. Les jeux de données ne proviennent pas directement de la station d’Al Hoceima.
2. `water_potability.csv` contient des valeurs manquantes et une cible modérément déséquilibrée.
3. `water_quality.xlsx` regroupe deux sources géographiques différentes.
4. La cible de corrosivité est dérivée de Larson et n’est pas une mesure indépendante de corrosion réelle.
5. L’Expérience A présente une dépendance structurelle entre prédicteurs et cible.
6. L’Expérience B reste une prédiction indirecte et ne remplace pas une validation chimique.
7. Langelier n’est pas calculé sans les paramètres nécessaires, notamment la température.
8. Une validation industrielle nécessiterait des données de la station, une séparation temporelle et idéalement une validation externe.

## 9. Tableau final pour le mémoire

```python
conclusion = pd.DataFrame({
    "Axe": [
        "Potabilité",
        "Agressivité A — avec Cl/SO4/HCO3",
        "Agressivité B — sans Cl/SO4/HCO3",
    ],
    "Cible": [
        "Potability (0/1)",
        "Larson_corrosive (IC >= 1)",
        "Larson_corrosive (IC >= 1)",
    ],
    "Rôle scientifique": [
        "Classification de potabilité",
        "Reproduction prédictive de la règle Larson",
        "Prédiction indirecte / analyse d'ablation",
    ],
    "Meilleur modèle": [None, None, None],
    "F1 test": [None, None, None],
    "ROC-AUC test": [None, None, None],
    "PR-AUC test": [None, None, None],
    "MCC test": [None, None, None],
})

if not summary.empty:
    mapping = {
        "Potabilité": "Potabilité",
        "Agressivité A — avec variables Larson": "Agressivité A — avec Cl/SO4/HCO3",
        "Agressivité B — sans variables Larson": "Agressivité B — sans Cl/SO4/HCO3",
    }
    for _, row in summary.iterrows():
        label = mapping.get(row["axe"])
        if label is None:
            continue
        mask = conclusion["Axe"].eq(label)
        conclusion.loc[mask, "Meilleur modèle"] = row.get("model")
        conclusion.loc[mask, "F1 test"] = row.get("f1")
        conclusion.loc[mask, "ROC-AUC test"] = row.get("roc_auc")
        conclusion.loc[mask, "PR-AUC test"] = row.get("pr_auc")
        conclusion.loc[mask, "MCC test"] = row.get("mcc")

display(conclusion)
conclusion.to_csv(RESULTS / "tableau_conclusion.csv", index=False)
```

## 10. Conclusion générale

Le projet met en place un pipeline complet et reproductible pour l’évaluation de la potabilité et de l’agressivité chimique de l’eau. La méthodologie sépare explicitement la **sélection des modèles**, réalisée par validation croisée sur les données d’entraînement, de leur **évaluation finale**, effectuée ensuite sur un jeu de test indépendant.

La principale amélioration méthodologique pour l’agressivité est la séparation entre une Expérience A, où les variables de Larson sont disponibles, et une Expérience B, où elles sont retirées. Cette seconde expérience mesure un signal prédictif indirect et évite de présenter comme découverte du machine learning la simple reproduction d’une formule déterministe.

Les résultats doivent enfin être interprétés dans les limites des données disponibles : ils constituent une démonstration méthodologique applicable au contexte du dessalement, mais pas une validation opérationnelle directe sur la station d’Al Hoceima sans données locales de procédé et de laboratoire.
