# 01 — Exploration de la potabilité

## Objectif

Ce notebook décrit le jeu `water_potability.csv`, vérifie sa qualité, étudie la distribution de la cible `Potability` et prépare une version nettoyée **sans effectuer d’imputation globale**.

> **Principe méthodologique :** les valeurs manquantes des variables explicatives sont conservées à ce stade. Elles seront imputées à l’intérieur des pipelines de modélisation afin d’éviter une fuite d’information entre entraînement, validation et test.

## 1. Configuration et reproductibilité

```python
from pathlib import Path
import sys
import warnings

warnings.filterwarnings("ignore")

PROJECT_ROOT = (
    Path.cwd().resolve().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd().resolve()
)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
FIGURES = PROJECT_ROOT / "reports" / "figures"
RESULTS = PROJECT_ROOT / "reports" / "results"
MODELS = PROJECT_ROOT / "models"

for directory in [DATA_PROCESSED, FIGURES, RESULTS, MODELS]:
    directory.mkdir(parents=True, exist_ok=True)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RANDOM_STATE = 42
pd.set_option("display.max_columns", 100)
```

```python
from src.data_cleaning import (
    basic_cleaning,
    data_quality_report,
    load_potability,
    save_processed,
)
```

## 2. Chargement des données

On commence par vérifier les dimensions, les premières observations et la présence de la cible.

```python
df_raw = load_potability(DATA_RAW / "water_potability.csv")
print("Dimensions :", df_raw.shape)
display(df_raw.head())
assert "Potability" in df_raw.columns, "La cible Potability est absente."
```

### Interprétation

Les dimensions confirment le nombre d’observations et de variables disponibles. À ce stade, aucune transformation statistique n’est appliquée : l’objectif est d’observer les données telles qu’elles sont fournies.

## 3. Structure, types et qualité des données

```python
structure = pd.DataFrame({
    "colonne": df_raw.columns,
    "dtype": df_raw.dtypes.astype(str).values,
})
display(structure)

quality = data_quality_report(df_raw)
display(quality)
print("Nombre de doublons :", df_raw.duplicated().sum())
```

### Interprétation

Le rapport de qualité permet d’identifier les colonnes avec valeurs manquantes, leur proportion et le nombre de valeurs distinctes. Les doublons exacts seront supprimés lors du nettoyage minimal.

## 4. Analyse des valeurs manquantes

```python
missing = (
    df_raw.isna().sum()
    .to_frame("missing_n")
    .assign(missing_pct=lambda x: 100 * x["missing_n"] / len(df_raw))
    .sort_values("missing_pct", ascending=False)
)
display(missing[missing["missing_n"] > 0].round(2))

if (missing["missing_n"] > 0).any():
    ax = missing.loc[missing["missing_n"] > 0, "missing_pct"].sort_values().plot.barh(
        figsize=(8, 4),
        title="Pourcentage de valeurs manquantes",
    )
    ax.set_xlabel("Pourcentage (%)")
    plt.tight_layout()
    plt.show()
```

### Interprétation

Une suppression systématique des lignes incomplètes pourrait réduire inutilement la taille du jeu. L’approche retenue consiste donc à conserver ces observations et à intégrer une **imputation par médiane dans le pipeline scikit-learn**. Ainsi, chaque valeur d’imputation est apprise uniquement sur les données d’entraînement du fold courant.

## 5. Nettoyage minimal

```python
df = basic_cleaning(df_raw, target="Potability")
print("Dimensions après nettoyage minimal :", df.shape)
print("Doublons restants :", df.duplicated().sum())
```

### Interprétation

Le nettoyage minimal supprime les doublons et garantit des colonnes numériques. Il ne remplit pas les valeurs manquantes des prédicteurs.

## 6. Analyse de la cible `Potability`

```python
counts = df["Potability"].value_counts().sort_index()
percentages = (100 * df["Potability"].value_counts(normalize=True).sort_index()).round(2)
class_table = pd.DataFrame({"effectif": counts, "pourcentage": percentages})
display(class_table)

ax = counts.plot.bar(figsize=(6, 4), title="Répartition de la cible Potability")
ax.set_xlabel("Classe")
ax.set_ylabel("Effectif")
plt.tight_layout()
plt.show()
```

### Interprétation

La distribution de la cible doit être comparée à une baseline naïve. Une classe majoritaire peut produire une accuracy non négligeable sans réelle capacité prédictive. C’est pourquoi le notebook 03 inclura un `DummyClassifier` et privilégiera également F1, ROC-AUC, PR-AUC et MCC.

## 7. Statistiques descriptives

```python
display(df.describe().T)
```

### Interprétation

Les statistiques descriptives permettent de repérer les différences d’échelle entre variables, ce qui justifie la standardisation pour les modèles sensibles à l’échelle, notamment la régression logistique et le SVM.

## 8. Distribution des variables

```python
features = [column for column in df.columns if column != "Potability"]
for column in features:
    ax = df[column].dropna().plot.hist(
        bins=30,
        figsize=(6, 3),
        title=f"Distribution — {column}",
    )
    ax.set_xlabel(column)
    plt.tight_layout()
    plt.show()
```

### Interprétation

Les histogrammes servent à repérer asymétries, concentrations et queues de distribution. Ils ne justifient pas automatiquement la suppression des valeurs extrêmes : dans un problème de qualité de l’eau, une valeur rare peut représenter une situation physico-chimique réelle.

## 9. Valeurs atypiques

```python
for column in features:
    fig, ax = plt.subplots(figsize=(6, 2.5))
    ax.boxplot(df[column].dropna(), vert=False)
    ax.set_title(f"Boxplot — {column}")
    ax.set_xlabel(column)
    plt.tight_layout()
    plt.show()
```

### Interprétation

Les boxplots sont utilisés ici comme outil descriptif. Les points éloignés ne sont pas supprimés automatiquement, car leur caractère « anormal » doit être confirmé par le contexte métier ou par des règles de qualité de mesure.

## 10. Corrélations linéaires

```python
corr = df.corr(numeric_only=True)
fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(corr, vmin=-1, vmax=1)
ax.set_xticks(range(len(corr)), corr.columns, rotation=90)
ax.set_yticks(range(len(corr)), corr.index)
fig.colorbar(im, ax=ax, label="Corrélation de Pearson")
ax.set_title("Matrice de corrélation")
plt.tight_layout()
plt.show()

display(corr[["Potability"]].sort_values("Potability", ascending=False))
```

### Interprétation

Une faible corrélation linéaire avec `Potability` ne signifie pas qu’une variable est inutile : les modèles non linéaires peuvent exploiter des interactions et des seuils qui ne sont pas visibles dans une corrélation de Pearson.

## 11. Comparaison descriptive par classe

```python
display(df.groupby("Potability")[features].median().T)
```

### Interprétation

Cette comparaison fournit une première lecture des différences de tendance centrale entre les deux classes. Elle reste descriptive et ne remplace pas l’évaluation hors échantillon réalisée dans le notebook 03.

## 12. Export des données nettoyées

```python
output_path = DATA_PROCESSED / "potability_clean.csv"
save_processed(df, output_path)
print("Fichier exporté :", output_path)
```
