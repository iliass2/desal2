# 03 — Modélisation de la potabilité

## Objectif

Construire une comparaison reproductible entre une baseline naïve et plusieurs modèles de classification afin de prédire `Potability`.

Le protocole est : **split train/test stratifié → validation croisée sur le train → choix du modèle → évaluation unique sur le test → interprétabilité**.

## 1. Configuration

```python
from pathlib import Path
import sys
import warnings

warnings.filterwarnings("ignore")

PROJECT_ROOT = (
    Path.cwd().resolve().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd().resolve()
)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
FIGURES = PROJECT_ROOT / "reports" / "figures"
RESULTS = PROJECT_ROOT / "reports" / "results"
MODELS = PROJECT_ROOT / "models"

for directory in [DATA_PROCESSED, FIGURES, RESULTS, MODELS]:
    directory.mkdir(parents=True, exist_ok=True)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RANDOM_STATE = 42
pd.set_option("display.max_columns", 100)
```

```python
from sklearn.model_selection import train_test_split
import joblib

from src.data_cleaning import basic_cleaning, load_potability
from src.evaluation import (
    classification_metrics,
    cross_validation_table,
    save_confusion_matrix,
    save_precision_recall_curve,
    save_roc_curve,
)
from src.interpretability import permutation_importance_table, shap_global_importance, shap_values_for_pipeline
from src.modeling import classification_models, needs_scaling
from src.preprocessing import make_model_pipeline
```

## 2. Préparation de `X` et `y`

```python
df = basic_cleaning(
    load_potability(DATA_RAW / "water_potability.csv"),
    target="Potability",
)

X = df.drop(columns="Potability")
y = df["Potability"].astype(int)

print("Dimensions X :", X.shape)
display(y.value_counts().rename("effectif").to_frame())
display((100 * y.value_counts(normalize=True)).round(2).rename("pourcentage").to_frame())
```

### Interprétation

La classe majoritaire fournit une référence naïve importante. Un modèle ne doit pas être jugé uniquement sur son accuracy : il doit apporter un gain crédible par rapport au `DummyClassifier` et conserver un bon compromis entre précision et rappel.

## 3. Séparation train/test stratifiée

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=RANDOM_STATE,
)

print("Train :", X_train.shape, "Test :", X_test.shape)
print("Distribution train :")
display(y_train.value_counts(normalize=True).round(4).to_frame("proportion"))
print("Distribution test :")
display(y_test.value_counts(normalize=True).round(4).to_frame("proportion"))
```

### Interprétation

La stratification conserve approximativement la même proportion de classes dans le train et le test. Le jeu de test reste isolé jusqu’à l’évaluation finale.

## 4. Modèles et pipelines

```python
models = classification_models(RANDOM_STATE)

pipelines = {}
for name, model in models.items():
    pipelines[name] = make_model_pipeline(
        model=model,
        numeric_features=X.columns,
        scale=needs_scaling(name),
    )

list(pipelines.keys())
```

### Interprétation

L’imputation médiane est intégrée à chaque pipeline. La standardisation est activée pour la régression logistique et le SVM, mais elle n’est pas nécessaire pour les modèles d’arbres. Le `DummyClassifier` sert de baseline.

## 5. Validation croisée sur le jeu d’entraînement

```python
cv_rows = []
for name, pipeline in pipelines.items():
    cv_table = cross_validation_table(
        pipeline,
        X_train,
        y_train,
        n_splits=5,
        random_state=RANDOM_STATE,
    )
    for _, row in cv_table.iterrows():
        cv_rows.append({"model": name, **row.to_dict()})

cv_results = pd.DataFrame(cv_rows)
cv_pivot = cv_results.pivot(index="model", columns="metric", values="mean")
display(cv_pivot.sort_values(["f1", "roc_auc"], ascending=False).round(4))
```

### Interprétation

Le tableau de validation croisée permet de comparer les modèles **uniquement sur le jeu d’entraînement**. La sélection du modèle final est effectuée à partir du F1 moyen, puis du ROC-AUC, du PR-AUC et du MCC comme critères de départage. Le jeu de test reste donc totalement indépendant de cette décision.

## 6. Sélection du meilleur modèle à partir de la validation croisée

```python
cv_pivot = cv_results.pivot(index="model", columns="metric", values="mean")

non_dummy_cv = cv_pivot.drop(index="Dummy Classifier", errors="ignore").copy()
ranking_metrics = ["f1", "roc_auc", "pr_auc", "mcc"]

cv_ranking = non_dummy_cv.sort_values(ranking_metrics, ascending=False)
best_name = cv_ranking.index[0]
best_model = pipelines[best_name]

print("Meilleur modèle sélectionné par validation croisée :", best_name)
display(cv_ranking.round(4))
```

### Interprétation

Le modèle retenu est choisi **avant toute consultation du jeu de test**. Cette séparation évite d’utiliser le test comme outil de sélection et préserve son rôle d’évaluation finale indépendante.

## 7. Entraînement final et évaluation unique sur le jeu de test

```python
best_model.fit(X_train, y_train)

final_metrics = {
    "model": best_name,
    **classification_metrics(best_model, X_test, y_test),
}
test_results = pd.DataFrame([final_metrics])

print("Performance finale sur le jeu de test :")
display(test_results.round(4))
```

### Interprétation

Le jeu de test est utilisé ici pour la première fois pour mesurer la capacité de généralisation du modèle déjà sélectionné. Les métriques finales ne servent plus à choisir entre plusieurs algorithmes : elles constituent l’estimation finale de performance.

## 8. Matrices de confusion et courbes

```python
from IPython.display import Image, display

cm_path = FIGURES / "potability_confusion_matrix.png"
cm_norm_path = FIGURES / "potability_confusion_matrix_normalized.png"
roc_path = FIGURES / "potability_roc_curve.png"
pr_path = FIGURES / "potability_precision_recall_curve.png"

save_confusion_matrix(best_model, X_test, y_test, f"Potabilité — {best_name}", cm_path)
save_confusion_matrix(
    best_model,
    X_test,
    y_test,
    f"Potabilité normalisée — {best_name}",
    cm_norm_path,
    normalize="true",
)
save_roc_curve(best_model, X_test, y_test, f"Potabilité — {best_name}", roc_path)
save_precision_recall_curve(best_model, X_test, y_test, f"Potabilité — {best_name}", pr_path)

for path in [cm_path, cm_norm_path, roc_path, pr_path]:
    display(Image(filename=str(path)))
```

### Interprétation

La matrice brute montre le nombre d’erreurs, tandis que la version normalisée facilite la comparaison entre classes. La courbe Precision-Recall est particulièrement utile lorsque la classe positive est moins fréquente.

## 9. Interprétabilité par permutation

```python
importance = permutation_importance_table(
    best_model,
    X_test,
    y_test,
    scoring="f1",
)
display(importance)

ax = importance.sort_values("importance_mean").plot.barh(
    x="feature",
    y="importance_mean",
    xerr="importance_std",
    figsize=(8, 5),
    legend=False,
    title=f"Permutation importance — {best_name}",
)
ax.set_xlabel("Baisse moyenne du F1 après permutation")
plt.tight_layout()
plt.show()
```

### Interprétation

Une importance par permutation élevée signifie que mélanger cette variable dégrade davantage le F1. Cela mesure une dépendance prédictive du modèle, mais ne démontre pas une relation causale.

## 10. SHAP — analyse complémentaire

```python
shap_table = pd.DataFrame()
try:
    if best_name in {"Random Forest", "XGBoost"}:
        sample = X_test.sample(min(150, len(X_test)), random_state=RANDOM_STATE)
        _, shap_explanation = shap_values_for_pipeline(best_model, sample)
        shap_table = shap_global_importance(shap_explanation)
        display(shap_table.head(15))
    else:
        print("SHAP n'est pas exécuté automatiquement pour ce type de meilleur modèle ; la permutation importance reste la référence principale.")
except Exception as exc:
    print("SHAP non disponible pour ce pipeline :", exc)
```

### Interprétation

SHAP est utilisé comme analyse complémentaire lorsqu’il est compatible avec le meilleur modèle. La conclusion du PFE ne doit pas dépendre d’un seul outil d’explicabilité.

## 11. Sauvegarde des résultats

```python
cv_results.to_csv(RESULTS / "potability_cv_metrics.csv", index=False)
test_results.to_csv(RESULTS / "potability_test_metrics.csv", index=False)
importance.to_csv(RESULTS / "potability_feature_importance.csv", index=False)
if not shap_table.empty:
    shap_table.to_csv(RESULTS / "potability_shap_importance.csv", index=False)

joblib.dump(best_model, MODELS / "potability_best_model.pkl")
print("Résultats et modèle sauvegardés.")
```

## 12. Conclusion du notebook

- Une baseline naïve est comparée aux modèles appris pendant la validation croisée.
- L’imputation et la standardisation sont confinées au pipeline.
- Le meilleur algorithme est sélectionné **uniquement à partir de la validation croisée sur le train**.
- Le jeu de test est ensuite utilisé une seule fois pour l’évaluation finale du modèle retenu.
- F1, ROC-AUC, PR-AUC et MCC complètent l’accuracy afin d’éviter une lecture trompeuse des performances.
- L’interprétabilité est étudiée sans confondre importance prédictive et causalité.
